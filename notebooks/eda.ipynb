```python
import pandas as pd, matplotlib.pyplot as plt
from PIL import Image
df = pd.read_csv("../data/audit.csv"); df = df[~df.corrupt]
fig, axes = plt.subplots(2, 6, figsize=(15, 5))
for r, cls in enumerate(["ai", "real"]):
    for c, p in enumerate(df[df["class"] == cls].sample(6, random_state=1).path):
        axes[r, c].imshow(Image.open("../" + p).convert("RGB")); axes[r, c].axis("off")
        axes[r, c].set_title(cls)
plt.tight_layout(); plt.savefig("../report_screenshots/1.3_grid.png", dpi=120)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
for ax, col in zip(axes, ["width", "height", "size_kb"]):
    for cls in ["ai", "real"]:
        ax.hist(df[df["class"] == cls][col], bins=40, alpha=0.6, label=cls)
    ax.set_title(col); ax.legend()
plt.tight_layout(); plt.savefig("../report_screenshots/1.3_hist.png", dpi=120)
```
